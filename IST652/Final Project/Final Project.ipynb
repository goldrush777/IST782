```python
players = pd.read_csv('player_splits.csv')
goalies = pd.read_csv('goalie_splits.csv')
```

```python
players.head()
```

```python
players['Total_Goals'] = players.groupby('Player')['G'].transform('sum')
players['Total_Points'] = players.groupby('Player')['PTS'].transform('sum')
players['Total_GP'] = players.groupby('Player')['GP'].transform('sum')
```

```python
players.head()
```

```python
players['Goals_Perc'] = players['G'] / players['Total_Goals']
players['Points_Perc'] = players['PTS'] / players['Total_Points']
```

```python
#Data Cleaning

#Removing Players with less than 100 career games played
players = players[players['Total_GP'] >= 100]

#Removing rows where GP is less than 15
players = players[players['GP'] >= 15]

players.head()
```

```python
idx = players.groupby('Opponent')['Points_Perc'].idxmax()

top_points_perc = players.loc[idx, ['Opponent', 'Player', 'Points_Perc']].reset_index(drop=True)

top_points_perc.head(20)
```

```python
idx = players.groupby('Opponent')['Goals_Perc'].idxmax()

top_goals_perc = players.loc[idx, ['Opponent', 'Player', 'Goals_Perc']].reset_index(drop=True)

top_goals_perc.head(20)
```

```python
idx = players.groupby('Opponent')['G'].idxmax()

top_goals_perc = players.loc[idx, ['Opponent', 'Player', 'G']].reset_index(drop=True)

top_goals_perc.head(20)
```

```python
idx = players.groupby('Opponent')['PTS'].idxmax()

top_points_perc = players.loc[idx, ['Opponent', 'Player', 'PTS']].reset_index(drop=True)

top_points_perc.head(20)
```

```python
Avalanche = players[players['Opponent'] == 'Colorado Avalanche']
Rangers = players[players['Opponent'] == 'New York Rangers']
```

```python
goalies['Total_SA'] = goalies.groupby('Player')['SA'].transform('sum')
goalies['Total_SV'] = goalies.groupby('Player')['SV'].transform('sum')
goalies['Career_SV%'] = goalies['Total_SV'] / goalies['Total_SA']
goalies['Total_GP'] = goalies.groupby('Player')['GP'].transform('sum')
```

```python
#Data Cleaning

#Removing Goalies with less than 50 career games played
goalies = goalies[goalies['Total_GP'] >= 50]

#Removing rows where GP is less than 15
goalies = goalies[goalies['GP'] >= 5]

goalies.head(20)
```

```python
idx = goalies.groupby('Opponent')['SV%'].idxmax()

top_save_perc = goalies.loc[idx, ['Opponent', 'Player', 'SV%']].reset_index(drop=True)

top_save_perc.head(20)
```

```python
players2 = pd.read_csv('player_splits.csv')
mcdavid = players2[players2['Player'] == 'Connor McDavid']
mcdavid
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

fig, ax = plt.subplots()
line, = ax.plot([], [], lw=2)
text_box = ax.text(0.5, 1.05, '', transform=ax.transAxes, ha='center', fontsize=12, bbox=dict(facecolor='white', alpha=0.5))

# Set axis labels
ax.set_xlabel('Games Played')
ax.set_ylabel('Goals Scored')

# Set axis limits (adjust if needed)
ax.set_xlim(0, mcdavid['GP'].max() + 5)
ax.set_ylim(0, mcdavid['G'].max() + 2)

# Initialization function
def init():
    line.set_data([], [])
    text_box.set_text('')
    return line, text_box

# Animation function
def animate(i):
    row = mcdavid.iloc[i]
    x = [0, row['GP']]
    y = [0, row['G']]
    line.set_data(x, y)
    text_box.set_text(f"Opponent: {row['Opponent']}")
    return line, text_box

# Create animation
ani = animation.FuncAnimation(fig, animate, init_func=init,
                              frames=len(mcdavid), interval=1000, blit=True)

plt.close(fig)
# ani.save('/content/drive/MyDrive/velo.gif', writer='pillow')
HTML(ani.to_jshtml())
```
