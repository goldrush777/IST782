```python
from urllib import request
from bs4 import BeautifulSoup
import pandas as pd
```

```python
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
url = "https://www.basketball-reference.com/leagues/NBA_2025_per_game.html"
```

```python
response = request.urlopen(url)
html = response.read().decode('utf8')
```

```python
htmlsoup = BeautifulSoup(html, 'html.parser')
```

```python
df_list = pd.read_html(str(htmlsoup))
player_stats = df_list[0]
player_stats = player_stats[player_stats['Player'] != 'Player'].reset_index(drop=True)
print(player_stats.head())
```

```python
standingsurl = "https://www.basketball-reference.com/leagues/NBA_2025_standings.html"
```

```python
response = request.urlopen(standingsurl)
html = response.read().decode('utf8')
```

```python
htmlsoup = BeautifulSoup(html, 'html.parser')
```

```python
df_list = pd.read_html(str(htmlsoup))
east_standings = df_list[0]
west_standings = df_list[1]
east_standings.head()
```

```python
west_standings.head()
```

```python
east_abbreviations = ['CLE', 'BOS', 'NYK', 'IND', 'MIL', 'DET', 'ORL', 'ATL', 'CHI', 'MIA', 'TOR', 'BKN', 'PHI', 'CHO', 'WAS']

east_standings['abbreviation'] = east_abbreviations

east_standings.head()
```

```python
west_abbreviations = ['OKC', 'HOU', 'LAL', 'DEN', 'LAC', 'MIN', 'GSW', 'MEM', 'SAC', 'DAL', 'PHO', 'POR', 'SAS', 'NOP', 'UTA']

west_standings['abbreviation'] = west_abbreviations

west_standings.head()
```

```python
east_standings = east_standings.rename(columns={"Eastern Conference": "Team"})
west_standings = west_standings.rename(columns={"Western Conference": "Team"})
```

```python
full_standings = pd.concat([east_standings, west_standings], ignore_index=True)

full_standings.head(20)
```

```python
player_stats = player_stats.rename(columns={"Team": "abbreviation"})
```

```python
full_df = player_stats.merge(full_standings, on='abbreviation', how='left')
full_df.head(10)
```

```python
#Data Cleaning

#Removing Players with less than 10 minutes played per game
full_df = full_df[full_df['MP'] > 10]

#Remvoing "2TM" Players by removing NAs in the Team column
full_df = full_df.dropna(subset=['Team'])
full_df.head(10)

full_df.to_csv('Full_NBA_Stats.csv', index=False)
```

```python
#League Leaders in Points, Assists, Rebounds, Field Goal %
full_df.nlargest(10, 'PTS')[['Player', 'Team', 'PTS']]
full_df.nlargest(10, 'TRB')[['Player', 'Team', 'TRB']]
full_df.nlargest(10, 'AST')[['Player', 'Team', 'AST']]
full_df.nlargest(10, 'FG%')[['Player', 'Team', 'FG%']]
```

```python
#Exporting League Leaders as a Single CSV

pts_leaders = full_df.nlargest(10, 'PTS')[['Player', 'Team', 'PTS']]
pts_leaders['Category'] = 'Points'

trb_leaders = full_df.nlargest(10, 'TRB')[['Player', 'Team', 'TRB']]
trb_leaders['Category'] = 'Rebounds'

ast_leaders = full_df.nlargest(10, 'AST')[['Player', 'Team', 'AST']]
ast_leaders['Category'] = 'Assists'

fg_leaders = full_df.nlargest(10, 'FG%')[['Player', 'Team', 'FG%']]
fg_leaders['Category'] = 'Field Goal Percentage'

#Normalizing the Stat Column Name since Category will show which Stat is being shown
pts_leaders = pts_leaders.rename(columns={'PTS': 'Stat'})
trb_leaders = trb_leaders.rename(columns={'TRB': 'Stat'})
ast_leaders = ast_leaders.rename(columns={'AST': 'Stat'})
fg_leaders = fg_leaders.rename(columns={'FG%': 'Stat'})


league_leaders = pd.concat([pts_leaders, trb_leaders, ast_leaders, fg_leaders], ignore_index=True)
league_leaders

league_leaders.to_csv('league_leaders.csv', index=False)
```

```python
#Top NBA Teams by Points per Game and Points Allowed per Game
full_standings.nlargest(10, 'PS/G')[['Team', 'PS/G']]
full_standings.nsmallest(10, 'PA/G')[['Team', 'PA/G']]
```

```python
#I'm using the same Process as Player League Leaders for Exporting Team League Leaders in a single CSV
PS_leaders = full_standings.nlargest(10, 'PS/G')[['Team', 'PS/G']]
PS_leaders['Category'] = 'Points Scored Per Game'

PA_leaders = full_standings.nsmallest(10, 'PA/G')[['Team', 'PA/G']]
PA_leaders['Category'] = 'Points Allowed Per Game'

PS_leaders = PS_leaders.rename(columns={'PS/G': 'Stat'})
PA_leaders = PA_leaders.rename(columns={'PA/G': 'Stat'})


team_league_leaders = pd.concat([PS_leaders, PA_leaders], ignore_index=True)
team_league_leaders

team_league_leaders.to_csv('team_league_leaders.csv', index=False)
```

```python
#Statistics for a Team of your Choice, Will Create a CSV file automatically
team = input("Please enter an NBA team (abbreviation) ")
individual_team_stats = full_df[full_df['abbreviation'] == team]
individual_team_stats.sort_values("PTS", ascending = False)

filename = f"{team}_stats.csv"
individual_team_stats.to_csv(filename, index=False)
```

```python
#Full Stats table for Teams' Most Popular Starters, and Everyone else
starters = full_df.groupby('abbreviation', group_keys=False).apply(lambda x: x.nlargest(5, 'GS'))
non_starters = full_df.groupby('abbreviation', group_keys=False).apply(lambda x: x[x['GS'].rank(method='first', ascending=False) > 5])
starters.head(20)
```

```python
#Summary Scoring Stats for Starters vs Non Starters Leaguewide
cols = ['PTS','FGA','FG%']
starters[cols].describe()
non_starters[cols].describe()
```

```python
#Saving Summary Stats as CSVs
starters_SS = starters[cols].describe()
non_starters_SS = non_starters[cols].describe()

starters_SS.to_csv("starters_summary_stats.csv", index=False)
non_starters_SS.to_csv("non_starters_summary_stats.csv", index=False)
```
